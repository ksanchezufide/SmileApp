```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image, ImageOps
```

```python
def cargar_imagen(ruta):

    img = Image.open(ruta)

    img = ImageOps.exif_transpose(img)
    img = img.convert("L")
    img = img.resize((256, 256))

    img = np.array(img) / 255.0

    return img
```

```python
ruta_train = '/content/drive/MyDrive/DatasetSingLenguaje/Train'
```

```python
datos = []

for letra in os.listdir(ruta_train):

    carpeta = os.path.join(ruta_train, letra)

    if os.path.isdir(carpeta):

        for archivo in os.listdir(carpeta):

            if archivo.lower().endswith((
    '.png', '.jpg', '.jpeg', '.bmp', '.gif', '.tiff', '.tif',
    '.webp', '.ppm', '.pgm', '.pbm', '.ico'
)):

                ruta_img = os.path.join(carpeta, archivo)

                try:

                    img = cargar_imagen(ruta_img)

                    brillo_promedio = np.mean(img)
                    desviacion_brillo = np.std(img)

                    pixeles_claros = np.sum(img > 0.7)
                    pixeles_oscuros = np.sum(img < 0.3)

                    datos.append({
                        "letra": letra,
                        "brillo_promedio": brillo_promedio,
                        "desviacion_brillo": desviacion_brillo,
                        "pixeles_claros": pixeles_claros,
                        "pixeles_oscuros": pixeles_oscuros
                    })

                except:
                    print("Imagen ignorada:", ruta_img)

df = pd.DataFrame(datos)

print(df.head())
```

```python
print(df["letra"].value_counts())
```

```python
print(df.describe())
```

```python
plt.hist(df["brillo_promedio"], bins=20)

plt.title("Histograma del brillo promedio")
plt.xlabel("Brillo")
plt.ylabel("Frecuencia")

plt.show()
```
